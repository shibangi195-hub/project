```python
name=input("Enter Your Name:")
print(f"Hey {name}! Welcome to IDBI Bank!!😊".center(50))
a="ATM"
print(a.center(50,"-"))
balance=50000.00
statement=(f"Hey {name} Your Initial Balance is: Rs. 50000.00")
print("1.Balance Check")
print("2.Cash Withdraw")
print("3.Balance Statement")
print()
choice=input("Choose from above options:")
if choice=="1":
    print(f"Hey {name}, Your current Balance is: Rs" +str(balance))
elif choice=="2":
    withdraw=float(input("Enter Amount:"))
    if withdraw<=balance:
        balance=balance-withdraw
        print("Withdrawl Successful".center(50,"-"))
        print("Remaining balance",str(balance))
        statement=statement+"\nWithdrew: Rs"+str(withdraw)+"\nNew Balance"+str(balance)
    else:
        print("Insufficient Balance!! Transaction Failed!!")
elif choice=="3":
    print("Account Statement".center(50,"-"))
    print(statement)
else:
    print("Invalid Option Selected!! Please Try Again")
```
